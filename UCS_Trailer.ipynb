Goal: find the cheapest path for a trailer from Nairobi to Kisumu on the highway network from Part 1.

## Step 1: Directed graph from Part 1
Edge weights are road distances in km (from Google Maps, rounded).

```python
graph = {
    "Nairobi":    {"Thika": 45, "Naivasha": 93},
    "Thika":      {"Nyeri": 113},
    "Nyeri":      {"Nyahururu": 104},
    "Nyahururu":  {"Nakuru": 68},
    "Naivasha":   {"Nakuru": 70},
    "Nakuru":     {"Kericho": 108, "Eldoret": 171},
    "Eldoret":    {"Kakamega": 106, "Kericho": 140},
    "Kericho":    {"Kisumu": 84},
    "Kakamega":   {"Kisumu": 53},
    "Kisumu":     {},
}
start, goal = "Nairobi", "Kisumu"
```

## Step 2: UCS implementation
A priority queue (frontier) always expands the node with the lowest path cost so far.

```python
import heapq

def ucs(graph, start, goal):
    frontier = [(0, start, [start])]   # (cost, node, path)
    explored = set()
    step = 0
    while frontier:
        cost, node, path = heapq.heappop(frontier)
        step += 1
        print(f"Step {step}: expand {node} (cost {cost})")
        if node == goal:
            print("Goal reached!")
            return path, cost
        if node in explored:
            print("   already explored, skip")
            continue
        explored.add(node)
        for nbr, w in graph[node].items():
            if nbr not in explored:
                heapq.heappush(frontier, (cost + w, nbr, path + [nbr]))
                print(f"   add {nbr} with cost {cost + w}")
        print("   frontier:", sorted((c, n) for c, n, _ in frontier))
    return None, float("inf")
```

## Step 3: Run the search

```python
path, cost = ucs(graph, start, goal)
print()
print("Optimal path:", " -> ".join(path))
print("Total cost  :", cost)
```
